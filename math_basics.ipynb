```python
nums = [1, 2, 3, 4, 5]
nums
```

```python
print(7 + 3)   # addition
print(7 - 3)   # subtraction
print(7 * 3)   # multiplication
print(7 / 3)   # division
print(7 // 3)  # whole-number division
print(7 % 3)   # remainder
print(7 ** 2)  # power
```

```python
total_with_sum = sum(nums)

total_with_loop = 0
for number in nums:
    total_with_loop += number

from functools import reduce
total_with_reduce = reduce(lambda running_total, number: running_total + number, nums, 0)
```

```python
mean = sum(nums) / len(nums)
smallest = min(nums)
largest = max(nums)

print(f"Mean: {mean}")         # 3.0
print(f"Smallest: {smallest}") # 1
print(f"Largest: {largest}")   # 5
```

```python
squares = [number ** 2 for number in nums]
sum_of_squares = sum(squares)

print(squares)         # [1, 4, 9, 16, 25]
print(sum_of_squares)  # 55
```

```python
a = 3
b = 4
c = 19

x = (c - b) / a
print(x)  # 5.0
print(a * x + b)  # Check: 19.0
```
