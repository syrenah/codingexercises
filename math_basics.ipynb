```python
mean = sum(nums) / len(nums)
smallest = min(nums)
largest = max(nums)

print(f"Mean: {mean}")         # 3.0
print(f"Smallest: {smallest}") # 1
print(f"Largest: {largest}")   # 5
```

```python
squares = [number ** 2 for number in nums]
sum_of_squares = sum(squares)

print(squares)         # [1, 4, 9, 16, 25]
print(sum_of_squares)  # 55
```

```python
def smallest_missing_positive(my_array):
    """
    Finds the smallest positive integer missing from the list A.
    Time complexity: O(n)
    Space complexity: O(n)
    """
    # Use a set for O(1) lookups
    positives = {x for x in my_array if x > 0}

    # Start checking from 1 upwards
    smallest = 1
    while smallest in positives:
        smallest += 1
    return smallest


if __name__ == "__main__":
    try:
        # Example input
        raw_input = "1 44 2  3 4 -4"
        if not raw_input:
            raise ValueError("Input cannot be empty.")

        # Convert to list of integers
        my_array = list(map(int, raw_input.split()))
        my_array = list(map(float, raw_input.split())) # nothing is whitespace
        # Compute result
        result = smallest_missing_positive(my_array)
        print(f"Smallest missing positive integer: {result}")

    except ValueError as e:
        print(f"Error: {e}")
```
